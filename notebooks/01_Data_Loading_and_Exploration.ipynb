```python
import pandas as pd

df = pd.read_csv("../data/raw/01_fund_master.csv")

df.head()
```

```python
# Check missing values
df.isnull().sum()

# Remove duplicates
df = df.drop_duplicates()

# Check data types
df.dtypes

# Save cleaned dataset
df.to_csv(
    "../data/processed/dataset1_cleaned.csv",
    index=False
)
```

```python
df["category"].value_counts()
```

```python
import matplotlib.pyplot as plt

category_counts = df["category"].value_counts()

category_counts.plot(kind="bar")

plt.title("Number of Mutual Funds by Category")
plt.xlabel("Category")
plt.ylabel("Number of Funds")

plt.show()
```

```python
df["fund_house"].value_counts()
```

```python
import matplotlib.pyplot as plt

fund_house_counts = df["fund_house"].value_counts()

fund_house_counts.plot(kind="bar", figsize=(10,5))

plt.title("Number of Schemes by Fund House")
plt.xlabel("Fund House")
plt.ylabel("Number of Schemes")
plt.xticks(rotation=45)

plt.show()
```

```python
df["risk_category"].value_counts()
```

```python
risk_counts = df["risk_category"].value_counts()

risk_counts.plot(kind="bar")

plt.title("Distribution of Risk Categories")
plt.xlabel("Risk Category")
plt.ylabel("Number of Funds")

plt.show()
```
