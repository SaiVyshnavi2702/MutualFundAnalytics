```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

plt.plot(
    df["date"],
    df["close_value"]
)

plt.title("Benchmark Index Trend")

plt.xlabel("Date")

plt.ylabel("Index Value")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(7,4))

plt.hist(
    df["close_value"],
    bins=30
)

plt.title("Benchmark Index Distribution")

plt.xlabel("Index Value")

plt.ylabel("Frequency")

plt.show()
```

```python
avg = (
    df.groupby("index_name")["close_value"]
    .mean()
)

avg
```

```python
avg.plot(
    kind="bar",
    figsize=(8,4)
)

plt.title("Average Benchmark Index")

plt.xlabel("Benchmark")

plt.ylabel("Average Index Value")

plt.xticks(rotation=45)

plt.show()
```
