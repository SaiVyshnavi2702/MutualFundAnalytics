```python
import pandas as pd
```

```python
fund_master = pd.read_csv("../data/raw/01_fund_master.csv")

nav_history = pd.read_csv("../data/raw/02_nav_history.csv")

print("Fund Master Loaded")
print("NAV History Loaded")
```

```python
fund_master.columns
```

```python
nav_history.columns
```

```python
fund_codes = set(fund_master["amfi_code"])

nav_codes = set(nav_history["amfi_code"])


print("Total Fund Master Codes :", len(fund_codes))

print("Total NAV History Codes :", len(nav_codes))


missing_codes = fund_codes - nav_codes


print("\nMissing Codes:")


if len(missing_codes) == 0:
    print("No missing AMFI codes found.")
else:
    print(missing_codes)


print("\nValidation Completed")
```

AMFI Code Validation Summary

Fund Master and NAV History datasets were compared using AMFI scheme codes.

All AMFI codes from the Fund Master dataset were checked against the NAV History dataset.

No missing scheme codes were identified.

This confirms that the datasets are consistent and ready for further analysis.
