```python
top_return = (
    df.sort_values(
        by="return_5yr_pct",
        ascending=False
    )
    [["scheme_name","return_5yr_pct"]]
    .head(10)
)

top_return
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    top_return["scheme_name"],
    top_return["return_5yr_pct"]
)

plt.title("Top 10 Funds by 5 Year Return")
plt.xlabel("Fund")
plt.ylabel("5 Year Return (%)")

plt.xticks(rotation=90)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["std_dev_ann_pct"],
    df["return_5yr_pct"]
)

plt.xlabel("Risk (Standard Deviation %)")
plt.ylabel("5 Year Return %")

plt.title("Risk vs Return Analysis")

plt.show()
```

```python
category_return = (
    df.groupby("category")["return_5yr_pct"]
    .mean()
    .sort_values(ascending=False)
)

category_return
```

```python
plt.figure(figsize=(10,5))

category_return.plot(
    kind="bar"
)

plt.title("Average 5 Year Return by Category")

plt.xlabel("Category")
plt.ylabel("Average Return (%)")

plt.xticks(rotation=90)

plt.show()
```
