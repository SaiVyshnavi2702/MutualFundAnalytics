```python
import pandas as pd

df = pd.read_csv("../data/raw/02_nav_history.csv")

df.head()
```

```python
# Check missing values
df.isnull().sum()

# Remove duplicates
df = df.drop_duplicates()

# Check data types
df.dtypes

# Save cleaned dataset
df.to_csv(
    "../data/processed/dataset2_cleaned.csv",
    index=False
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))

plt.plot(df["date"], df["nav"])

plt.title("NAV Movement Over Time")
plt.xlabel("Date")
plt.ylabel("NAV Value")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(7,4))

plt.hist(df["nav"], bins=30)

plt.title("NAV Distribution")
plt.xlabel("NAV")
plt.ylabel("Frequency")

plt.show()
```
